```python
'''Exercise 9 – Activation Functions 
After calculating the weighted sum (Z), we pass it through an Activation Function.
We'll use the same output (Z) from the previous exercise.

Create Sample Values'''

import numpy as np

# Output from previous layer (before activation)
z = np.array([-3, -1, 0, 1, 3])

print("Weighted Sum (Z):")
print(z)
```

```python
'''Sigmoid Function'''
import numpy as np

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

sigmoid_output = sigmoid(z)

print("Sigmoid Output:")
print(sigmoid_output)
```

```python
'''ReLU Function'''
import numpy as np

def relu(x):
    return np.maximum(0, x)

relu_output = relu(z)

print("ReLU Output:")
print(relu_output)
```

```python
'''Tanh Function'''
import numpy as np

def tanh(x):
    return np.tanh(x)

tanh_output = tanh(z)

print("Tanh Output:")
print(tanh_output)
```

```python
'''Plot All Activation Functions'''
import numpy as np
import matplotlib.pyplot as plt

# Generate values
x = np.linspace(-5, 5, 200)

# Activation Functions
sigmoid = 1 / (1 + np.exp(-x))
relu = np.maximum(0, x)
tanh = np.tanh(x)

# Plot
plt.figure(figsize=(8,5))

plt.plot(x, sigmoid, label="Sigmoid")
plt.plot(x, relu, label="ReLU")
plt.plot(x, tanh, label="Tanh")

plt.title("Activation Functions")
plt.xlabel("Input (Z)")
plt.ylabel("Output")
plt.grid(True)
plt.legend()

plt.show()
```

```python
'''Apply Activation to the Previous Exercise'''
import numpy as np

# Input (2 students, 3 features)
X = np.array([
    [2, 5, 8],
    [4, 6, 3]
])

# Weights
W = np.array([
    [0.2, 0.5],
    [0.4, 0.3],
    [0.6, 0.8]
])

# Bias
b = np.array([1, 2])

# Weighted Sum (Z)
Z = np.dot(X, W) + b

print("Before Activation (Z):")
print(Z)

# ReLU Activation
A = np.maximum(0, Z)
B = 1 / (1 + np.exp(-Z))
C = np.tanh(Z)

print("\nAfter ReLU Activation:")
print(A)
print("\nAfter Sigmoid Activation:")
print(B)
print("\nAfter Tanh Activation:")
print(C)
```

```python
'''Step 2 – Create a Small Dataset
Let's predict Pass (1) or Fail (0) based on:
Study Hours
Attendance'''
import numpy as np

X = np.array([
    [2, 50],
    [3, 60],
    [4, 65],
    [5, 70],
    [6, 75],
    [7, 80],
    [8, 85],
    [9, 90]
])

y = np.array([
    0,
    0,
    0,
    0,
    1,
    1,
    1,
    1
])
```

```python
'''Step 3 – Build the Neural Network'''
model = Sequential([
    Dense(4, activation="relu", input_shape=(2,)),
    Dense(1, activation="sigmoid")
])

'''Explanation:
Input (2 Features)
↓
Hidden Layer (4 Neurons)
↓
Output Layer (1 Neuron)'''
```

```python
'''Step 4 – View the Model'''
model.summary()
```

```python
'''Step 5 – Compile the Model'''
model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)
```

```python
'''Step 6 – Train the Model'''
history = model.fit(
    X,
    y,
    epochs=100,
    verbose=1
)
```

```python
'''Step 7 – Make Predictions'''
prediction = model.predict(X)
print(prediction)
```

```python
'''These are probabilities.
Anything above 0.5 is classified as Pass.'''
```

```python
'''Step 8 – Predict a New Student'''
new_student = np.array([[7, 78]])

prediction = model.predict(new_student)

print(prediction)
```

```python
'''Exercise 2 – Model with Mean Squared Error'''
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

# Dataset
X = np.array([
    [2,50],
    [3,60],
    [4,65],
    [5,70],
    [6,75],
    [7,80],
    [8,85],
    [9,90]
])

y = np.array([0,0,0,0,1,1,1,1])

# Model
model_mse = Sequential([
    Dense(4, activation='relu', input_shape=(2,)),
    Dense(1, activation='sigmoid')
])

model_mse.compile(
    optimizer='adam',
    loss='mse',
    metrics=['accuracy']
)

history_mse = model_mse.fit(
    X,
    y,
    epochs=100,
    verbose=0
)

# Evaluate
loss_mse, acc_mse = model_mse.evaluate(X, y, verbose=0)

print("===== MSE MODEL =====")
print("Loss :", loss_mse)
print("Accuracy :", acc_mse)

```

```python
'''Make Predictions with SME '''
print("\nPredictions:")
print(model_mse.predict(X))
```

```python
'''Exercise 4 – build model with Gradient Descent'''
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.optimizers import SGD
import matplotlib.pyplot as plt

# Dataset
X = np.array([
    [2,50],
    [3,60],
    [4,65],
    [5,70],
    [6,75],
    [7,80],
    [8,85],
    [9,90]
], dtype=np.float32)

# Normalize
X[:,0] = X[:,0] / 9
X[:,1] = X[:,1] / 90

y = np.array([0,0,0,0,1,1,1,1], dtype=np.float32)

# Model
model = Sequential([
    Dense(4, activation='relu', input_shape=(2,)),
    Dense(1, activation='sigmoid')
])

# Gradient Descent Optimizer
optimizer = SGD(learning_rate=0.01)

model.compile(
    optimizer=optimizer,
    loss='mse',
    metrics=['accuracy']
)

history = model.fit(
    X,
    y,
    epochs=100,
    verbose=1
)
```

```python
plt.figure(figsize=(8,5))

plt.plot(history.history['loss'])

plt.title("Gradient Descent using MSE")

plt.xlabel("Epoch")

plt.ylabel("Loss")

plt.grid()

plt.show()

```
